```python
USER_INPUT = 'Paris'

from langchain_mistralai import ChatMistralAI

from langchain import PromptTemplate

llm = ChatMistralAI(model="mistral-large-latest",
    temperature=0,
    max_retries=2,
    # other params...)

template = """ I am travelling to {location}. What are the top 3 things I can do while I am there. Be very specific and respond as three bullet points """

prompt = PromptTemplate(

input_variables=["location"],

template=template,

)

final_prompt = prompt.format(location=USER_INPUT )

print(f"LLM Output: {llm(final_prompt)}")
```

```python
from langchain_mistralai import ChatMistralAI

# 1️⃣ Initialisation du modèle
llm = ChatMistralAI(
    model="mistral-small",
    temperature=0.7
)

# 2️⃣ Mémoire MANUELLE
history = []

# Message système
history.append({
    "role": "system",
    "content": "Tu es un assistant utile et concis."
})

print("💬 Chat avec Mistral (tape 'exit' pour quitter)\n")

# 3️⃣ Variable de contrôle de la boucle
running = True

while running:
    user_input = input("👤 Vous : ")

    # Condition d'arrêt
    if user_input.lower() in ["exit", "quit"]:
        print("👋 Bye")
        running = False
    else:
        # Ajouter le message utilisateur
        history.append({
            "role": "user",
            "content": user_input
        })

        # Appel au modèle
        response = llm(history)

        # Ajouter la réponse de l'IA
        history.append({
            "role": "assistant",
            "content": response.content
        })

        # Affichage
        print(f"🤖 Mistral : {response.content}\n")
```
