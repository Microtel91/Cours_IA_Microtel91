```python
from langchain_mistralai import ChatMistralAI

llm = ChatMistralAI(
    model="mistral-large-latest",
    temperature=0,
    api_key="Rentrez votre clée API."
)
```

```python
messages = [
    (
        "system",
        "Vous êtes un assistant utile qui traduit de l'anglais vers le français. Traduisez la phrase de l'utilisateur.",
    ),
    ("human", "J'adore programmer."),
]
ai_msg = llm.invoke(messages)
```

```python
print(ai_msg.content)
```
