## Boucles

### For

friends = ['Joseph', 'Glenn', 'Sally']
for friend in friends:
print('Happy New Year:', friend)
print('Done!')

total = 0
for itervar in [3, 41, 12, 9, 74, 15]:
total = total + itervar
print('Total: ', total)

### While

liste = [3, 41, 12, 9, 74, 15]
total = 0
indice = 0
while indice < 6:
    total = total + liste[indice]
    indice = indice + 1

n = 10
while True:
print(n, end=' ')
n = n - 1
print('Done!')
