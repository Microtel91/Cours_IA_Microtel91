# Amélioration LLM

### Juste parler à l’IA

```python
from langchain_mistralai import ChatMistralAI

llm = ChatMistralAI(model="mistral-large-latest",
                    api_key="Rentrez votre clée API.")

response = llm.invoke("Explique-moi en 3 phrases ce qu'est l'IA.")

print(response.content)
```

### L’utilisateur écrit le message

```python
from langchain_mistralai import ChatMistralAI

llm = ChatMistralAI(model="mistral-large-latest",
                    api_key="Rentrez votre clée API.")

message = input("Écris un message : ")

response = llm.invoke(message)

print("IA :", response.content)
```

### Faire une boucle (discussion continue)

```python
from langchain_mistralai import ChatMistralAI

llm = ChatMistralAI(model="mistral-large-latest",
                    api_key="Rentrez votre clée API.")


message = input("Toi : ")

while message != "quitter":

    response = llm.invoke(message)
    print("IA :", response.content)
    message = input("Toi : ")
```

### Ajouter la mémoire

```python
from langchain_mistralai import ChatMistralAI

llm = ChatMistralAI(model="mistral-large-latest",
                    api_key="Rentrez votre clée API.")

history = [
    {"role": "system", "content": "Tu es un enseignant qui s'adresse à des enfants de 8 ans."}
]

message = input("Toi : ")

while message != "quitter":

    history.append({"role": "user", "content": message})

    response = llm.invoke(history)

    history.append({"role": "assistant", "content": response.content})

    print("IA :", response.content)
    message = input("Toi : ")
```
